# Cell 2 — Figure 1: Confusion Matrix placeholder

Figure này tạo confusion matrix ước tính từ per-class F1. Đây không phải confusion matrix thực vì không có prediction/y_true theo mẫu.

```python
def confusion_from_f1(f1_neg, f1_neu, f1_pos):
    diag = np.clip(np.array([f1_neg, f1_neu, f1_pos], dtype=float), 0.02, 0.98)
    cm = np.zeros((3, 3), dtype=float)
    for i, d in enumerate(diag):
        cm[i, i] = d
        rem = 1.0 - d
        if i == 0:
            cm[i, 1] = rem * 0.65
            cm[i, 2] = rem * 0.35
        elif i == 1:
            cm[i, 0] = rem * 0.50
            cm[i, 2] = rem * 0.50
        else:
            cm[i, 0] = rem * 0.35
            cm[i, 1] = rem * 0.65
    return cm

teacher_f1_neu = teacher_mean["f1_neutral"]
teacher_f1_other = (3 * teacher_mean["f1_macro"] - teacher_f1_neu) / 2
tier1_direct = direct_summary_by_tier["Tier1"]
tier2_direct = direct_summary_by_tier["Tier2"]
tier2_twostage = two_stage_summary[("Tier2", 2)]
tier2_ts_other = (3 * tier2_twostage["f1_macro"] - tier2_twostage["f1_neutral"]) / 2

cms = {
    "Teacher": confusion_from_f1(teacher_f1_other, teacher_f1_neu, teacher_f1_other),
    "Tier1-Direct-spk4": confusion_from_f1(tier1_direct["f1_neg"], tier1_direct["f1_neu"], tier1_direct["f1_pos"]),
    "Tier2-Direct-spk4": confusion_from_f1(tier2_direct["f1_neg"], tier2_direct["f1_neu"], tier2_direct["f1_pos"]),
    "Tier2-Twostage-spk2": confusion_from_f1(tier2_ts_other, tier2_twostage["f1_neutral"], tier2_ts_other),
}

fig, axes = plt.subplots(2, 2, figsize=(11, 9))
for ax, (title, cm) in zip(axes.ravel(), cms.items()):
    sns.heatmap(cm * 100, annot=True, fmt=".1f", cmap="Blues", vmin=0, vmax=100,
                xticklabels=LABELS, yticklabels=LABELS, cbar=False, ax=ax)
    ax.set_title(title)
    ax.set_xlabel("Predicted label")
    ax.set_ylabel("True label")

fig.suptitle("Estimated Confusion Matrices (normalized by true label)", fontsize=14, y=1.02)
fig.text(0.5, -0.01, "Estimated from per-class F1, not actual confusion matrices.", ha="center", fontsize=11)
save_current_figure("fig1_confusion_placeholder.png")
plt.show()
```

**Nhận xét ngắn:** Hình 1 chỉ minh họa xu hướng collapse/khôi phục lớp neutral dựa trên per-class F1. Tier2 Direct KD có ô neutral rất thấp, trong khi Tier2 two-stage spk2 phục hồi rõ ở lớp này.

# Cell 3 — Figure 2: F1-neutral Direct KD vs Two-stage KD

Grouped bar chart so sánh Direct KD spk4 với Two-stage KD spk4 theo F1-neutral.

```python
rows = []
for tier in ["Tier1", "Tier2", "Tier3"]:
    direct_vals = [direct_kd[tier][seed]["f1_neu"] for seed in SEEDS]
    direct_mean, direct_std = mean_std(direct_vals)
    ts = two_stage_summary[(tier, 4)]
    rows.append({"tier": tier, "method": "Direct KD spk4", "mean": direct_mean, "std": direct_std})
    rows.append({"tier": tier, "method": "Two-stage KD spk4", "mean": ts["f1_neutral"], "std": ts["f1_neutral_std"]})
df_neu = pd.DataFrame(rows)

fig, ax = plt.subplots(figsize=(10, 5.5))
sns.barplot(data=df_neu, x="tier", y="mean", hue="method", palette=["#9ecae1", "#fb6a4a"], ax=ax)

for container, method in zip(ax.containers, ["Direct KD spk4", "Two-stage KD spk4"]):
    subset = df_neu[df_neu["method"] == method].reset_index(drop=True)
    for bar, (_, row) in zip(container, subset.iterrows()):
        x = bar.get_x() + bar.get_width() / 2
        y = bar.get_height()
        ax.errorbar(x, y, yerr=row["std"], color="black", capsize=4, linewidth=1)
        ax.text(x, y + 0.025, f"{y:.2f}", ha="center", va="bottom", fontsize=9)

teacher_neu = teacher_mean["f1_neutral"]
ax.axhline(teacher_neu, color=PALETTE["Teacher"], linestyle="--", linewidth=2,
           label=f"Teacher mean = {teacher_neu:.4f}")
ax.annotate("0.07 -> 0.56 (+0.49)", xy=(1.18, 0.561), xytext=(1.35, 0.70),
            arrowprops=dict(arrowstyle="->", color="black"), fontsize=11)
ax.set_title("F1-neutral: Direct KD vs Two-stage KD")
ax.set_xlabel("Tier")
ax.set_ylabel("F1-neutral")
ax.set_ylim(0, 0.8)
ax.legend(loc="upper left")
save_current_figure("fig2_f1_neutral_direct_vs_twostage.png")
plt.show()
```
